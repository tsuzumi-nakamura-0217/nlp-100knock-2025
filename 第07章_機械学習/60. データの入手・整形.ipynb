# 問題
GLUEのウェブサイトからSST-2データセットを取得せよ。学習データ（train.tsv）と検証データ（dev.tsv）のぞれぞれについて、ポジティブ (1) とネガティブ (0) の事例数をカウントせよ。

```python
def count(path):
  with open(path, 'r', encoding='utf-8') as f:
    count_0 = 0
    count_1 = 0
    for row in f:
      row = row.strip().split("\t")
      if not row:
          continue
      # カテゴリ宣言行
      if row[1] == "0":
        count_0 += 1
      elif row[1] == "1":
        count_1 += 1
      else:
        print("0と1以外です：", row[1])
    return count_0, count_1
```

```python
path_dev = "./SST-2/dev.tsv"
path_train = "./SST-2/train.tsv"
```

```python
count(path_train)
```
