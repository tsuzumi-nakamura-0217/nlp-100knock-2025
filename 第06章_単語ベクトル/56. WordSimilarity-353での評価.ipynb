# 問題
The WordSimilarity-353 Test Collectionの評価データをダウンロードし、単語ベクトルにより計算される類似度のランキングと、人間の類似度判定のランキングの間のスピアマン相関係数を計算せよ。

```python
from scipy.stats import spearmanr
from gensim.models import KeyedVectors

kv_path  = "./GoogleNews-vectors-negative300.bin"
file = './WordSimilarity 353/combined.csv'
kv = KeyedVectors.load_word2vec_format(kv_path, binary=True)
human = []
w2v = []
with open(file, 'r', encoding='utf-8') as f:
    next(f)
    for row in f:
        cols = row.rstrip().split(',')
        human.append(float(cols[2]))
        w2v.append(kv.similarity(cols[0], cols[1]))

correlation, pvalue = spearmanr(human, w2v)

print('スピアマン相関係数: {}'.format(correlation))
```
