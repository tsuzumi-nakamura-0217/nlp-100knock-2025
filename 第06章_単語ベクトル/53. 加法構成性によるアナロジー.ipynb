# 問題
“Spain”の単語ベクトルから”Madrid”のベクトルを引き、”Athens”のベクトルを足したベクトルを計算し、そのベクトルと類似度の高い10語とその類似度を出力せよ。

```python
from gensim.models import KeyedVectors
import numpy as np
def word2vector(word):
  path = "/Users/nakamuratuzumi/Jupyterlab/言語処理100本ノック 2025/第06章_単語ベクトル/GoogleNews-vectors-negative300.bin"

  kv = KeyedVectors.load_word2vec_format(path, binary=True, limit=None)

  key = word
  if key in kv:
    vec = kv[key] 
    print(vec.shape)
    print(vec[:10])
  else:
    print("キーが見つかりませんでした:", key)
  return vec

def vector_minus(vec_a, vec_b):
  minus_vector = vec_a - vec_b
  return minus_vector

def vector_add(vec_c, vec_d):
  add_vec = vec_c + vec_d
  return add_vec

def top10(word):
  path = "/Users/nakamuratuzumi/Jupyterlab/言語処理100本ノック 2025/第06章_単語ベクトル/GoogleNews-vectors-negative300.bin"
  kv = KeyedVectors.load_word2vec_format(path, binary=True, limit=None)
  similar_words = kv.most_similar("word", topn=10)

  for word, score in similar_words:
      print(f"{word}: {score:.4f}")
```

```python
spain = word2vector("Spain")
madrid = word2vector("Madrid")
athens = word2vector("Athens")
```

```python
vector = vector_minus(spain, madrid)
final_vector = vector_add(vector, athens)
```

```python
top10(final_vector)
```
