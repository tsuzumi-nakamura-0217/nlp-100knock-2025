# 問題
Google Newsデータセット（約1,000億単語）での学習済み単語ベクトル（300万単語・フレーズ、300次元）をダウンロードし、”United States”の単語ベクトルを表示せよ。ただし、”United States”は内部的には”United_States”と表現されていることに注意せよ。

```python
from gensim.models import KeyedVectors

path = "/Users/nakamuratuzumi/Jupyterlab/言語処理100本ノック 2025/第06章_単語ベクトル/GoogleNews-vectors-negative300.bin.gz"

kv = KeyedVectors.load_word2vec_format(path, binary=True, limit=None)

key = "United_States"
if key in kv:
    vec = kv[key]  # 同じく 300 次元
    print(vec.shape)
    print(vec[:10])
else:
    print("キーが見つかりませんでした:", key)
```
