54の実行結果を用い、意味的アナロジー（semantic analogy）と文法的アナロジー（syntactic analogy）の正解率を測定せよ。

### 意味的アナロジーと文法的アナロジー
.txtの: gramで始まるものが文法的アナロジー。それ以外で始まるのが意味的アナロジー。

```python
from gensim.models import KeyedVectors

def load_section(txt_path):
    syn = []  # syntactic
    sem = []  # semantic
    current = None  # "syn" or "sem"

    with open(txt_path, encoding="utf-8") as f:
        for raw in f:
            line = raw.strip()
            if not line:
                continue

            if line.startswith(":"):
                # セクション見出し
                header = line[1:].strip()
                if header.startswith("gram1"):  
                    current = "syn"
                elif header == "capital-common-countries":  # この1種類だけ sem
                    current = "sem"
                else:
                    current = None  # それ以外の意味系は無視
                continue

            # ここはデータ行（4語）
            parts = line.split()
            if len(parts) >= 4:
                quad = tuple(parts[:4])
                if current == "syn":
                    syn.append(quad)
                elif current == "sem":
                    sem.append(quad)
                # current が None のときは無視

    return syn, sem


def run_analogy(items, kv: KeyedVectors):
    rows = []
    for a, b, c, d in items:

        # a,b,c 自身が候補に来ることがあるので、除外して最初の候補を採用
        ban = {a, b, c}
        for pred_word, sim in kv.most_similar(positive=[b, c], negative=[a], topn=10):
            if pred_word not in ban:
                print(a)
                rows.append({
                    "a": a, "b": b, "c": c,
                    "pred_word": pred_word,
                    "true": d
                })
                break
        else:
            rows.append({
                "a": a, "b": b, "c": c,
                "pred_word": "",
                "true": d
            })
    return rows

# ==== 正解率 ====
def rate(rows):
  count = 0
  for row in rows:
    if row["pred_word"] == row["true"]:
      count +=1
  rate = count/len(rows)
  return rate
# ===== 実行 =====
txt_path = "単語ベクトルの質問.txt"
kv_path  = "/Users/nakamuratuzumi/Jupyterlab/言語処理100本ノック 2025/第06章_単語ベクトル/GoogleNews-vectors-negative300.bin"
syn, sem = load_section(txt_path)
kv = KeyedVectors.load_word2vec_format(kv_path, binary=True)
```

```python
syn_results = run_analogy(syn, kv)
```

```python
sem_results = run_analogy(sem, kv)
```
